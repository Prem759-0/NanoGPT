```python
# Step 3: Load Text-to-Video Spatio-Temporal Diffusion Model
import torch
from diffusers import DiffusionPipeline, DPMSolverMultistepScheduler
from diffusers.utils import export_to_video

pipe = DiffusionPipeline.from_pretrained(
    "damo-vilab/text-to-video-ms-1.7b",
    torch_dtype=torch.float16,
    variant="fp16"
)
pipe.scheduler = DPMSolverMultistepScheduler.from_config(pipe.scheduler.config)
pipe.enable_model_cpu_offload()
print("Video Diffusion Pipeline loaded successfully!")
```

```python
# Step 4: Generate High-Definition Video Clip (.mp4)
import os

prompt = "A cinematic drone flyover of a futuristic neon cyberpunk metropolis with flying cars, 4k, hyper-lapse"
negative_prompt = "low quality, blurry, distorted, static, jittery"

print(f"Generating video for prompt: '{prompt}'...")
video_frames = pipe(
    prompt=prompt,
    negative_prompt=negative_prompt,
    num_inference_steps=25,
    num_frames=16
).frames[0]

os.makedirs("generated_videos", exist_ok=True)
output_video_path = "generated_videos/ai_video_kling_caliber.mp4"
export_to_video(video_frames, output_video_path, fps=8)
print(f"Video generated successfully! Saved to: {output_video_path}")
```

```python
# Step 5: Preview Video inside Colab
from IPython.display import HTML
from base64 import b64encode

mp4 = open(output_video_path, 'rb').read()
data_url = "data:video/mp4;base64," + b64encode(mp4).decode()
HTML(f"""
<video width=500 controls autoplay loop>
    <source src="{data_url}" type="video/mp4">
</video>
""")
```

```python
# Step 6: Download MP4 Video to Local Computer
from google.colab import files
files.download(output_video_path)
print("Download started! You can drop this MP4 into your NanoGPT Lab generated directory.")
```
